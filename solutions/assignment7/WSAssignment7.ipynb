```python
#3. Write a function to read the file graph.dat and convert it to the adjacency matrix
import pandas as pd
import csv
import numpy

def plot():
    
    data = pd.read_csv('graph.dat', sep="\t")[1:]
    df = pd.DataFrame(data["% sym unweighted"].apply(lambda x: pd.Series(str(x).split(" ")))).apply(pd.to_numeric)
    graphDat = df.to_numpy()
    
    maxVal = graphDat.max()+1
    adjacencyMatrix = numpy.zeros((maxVal, maxVal))
    
    adjacencyMatrix[graphDat[:,0], graphDat[:,1]] = 1
    adjacencyMatrix[graphDat[:,1], graphDat[:,0]] = 1
    
    return adjacencyMatrix[1:,1:]
    
matrix = plot()
matrix
```

```python
#4. Write a function to compute the Gini coefficient g of the degrees of vertices in G.
# The function should have the adjacency matrix of G as its only parameter.

def gini(x):
    matrix = numpy.add.reduce(x)
    x = 0
    for i, j in enumerate(matrix[:-1], 1):
        x += numpy.sum(numpy.abs(j - matrix[i:]))
    giniCoef = x / (len(matrix)**2 * numpy.mean(matrix))
    
    return giniCoef
    
coef = gini(matrix)
coef
```

```python
#5. Write a function to compute the number of triangles t in the graph G. 
# The function should have the adjacency matrix of G as its only parameter.

def multiply(A, B, l):
    a = [[None] * l for i in range(l)]
    
    for i in range(l):
        for j in range(l):
            a[i][j] = 0
            
    for i in range(l):
        for j in range(l):
            a[i][j] = 0
            for k in range(l):
                a[i][j] += A[i][k] * B[k][j]
    return a
                
def triangles(matrix):
    l = len(matrix)
    c = multiply(matrix,matrix,l)
    d = multiply(matrix,c,l)
    
    num = 0
    for i in range(l):
        num += d[i][i]
    return num // 6

t = triangles(matrix)
t
```
